# READING DATA

```python
with open(r"C:\D\Data Science CodeWithHarry\finaldata.txt", encoding="utf-8") as f:
    data = f.read()

    

chunks=data.split("\n\n")
chunks=[c for c in chunks if len(c)>3]

# print(chunks[4])
```

# DATA PARSING

```python
def parse_chunk(chunk):
    chunk=chunk.strip()
    sep_chunk=chunk.split("\n")

    username=sep_chunk[0]

    posts=float(sep_chunk[1].split(" post")[0].replace(",","").replace("K","").replace("M",""))
    if "K" in sep_chunk[1]:
        posts=int(posts*1000)
    elif "M" in sep_chunk[1]:
        posts=int(posts*1000000)
    else:
        posts=int(posts)

    followers=float(sep_chunk[2].split(" follower")[0].replace(",","").replace("K","").replace("M",""))
    if "K" in sep_chunk[2]:
        followers=int(followers*1000)
    elif "M" in sep_chunk[2]:
        followers=int(followers*1000000)
    else:
        followers=int(followers)

    following=float(sep_chunk[3].split(" following")[0].replace(",","").replace("K","").replace("M",""))
    if "K" in sep_chunk[3]:
        following=int(following*1000)
    elif "M" in sep_chunk[3]:
        following=int(following*1000000)
    else:
        following=int(following)

    name=sep_chunk[4]

    if (len(sep_chunk)>5):
        type_of_page=sep_chunk[5]
        bio=sep_chunk[6:]
    else:
        type_of_page="Unknown"
        bio=[]
    

    user_dict={"username":username,"posts":posts,"followers":followers,"following":following,"name":name,"typeof_page":type_of_page,"bio":bio}

    return user_dict
```

```python
user_data=[]

for chunk in chunks:
    user=parse_chunk(chunk)
    user_data.append(user)

print(user_data)
```

# SAVING DATA

```python
import json

s=json.dumps(user_data,indent=4)
print(s)
# with open(r"C:\D\Data Science CodeWithHarry\finalData.txt", "w") as f:
#     f.write(s)

with open(r"C:\D\Data Science CodeWithHarry\finalData.json", "w") as f:
    json.dump(user_data,f,indent=4)
```

```python
with open(r"C:\D\Data Science CodeWithHarry\ finalData.txt", "w", encoding="utf-8") as file:
    for person in user_data:
        file.write(f"\n{person['name']}\n")
        file.write(f"{person['posts']}\n")
        file.write(f"{person['followers']}\n")
        file.write(f"{person['following']}\n")
        file.write(f"{person['typeof_page']}\n")

        for line in person["bio"]:
            file.write(f"{line}\n")
```

# FINDING MAX POSTS AND FOLLOWERS

```python
# Finding MAximum Posts
max_posts=max(user_data,key=lambda x:x["posts"])
print(f"{max_posts['username']} have maximum posts i.e. {max_posts['posts']}")

# Finding MAximum Followers
max_followers=max(user_data,key=lambda x:x["followers"])
print(f"{max_followers['username']} have maximum followers i.e {max_followers['followers']}")
```
